```kotlin
fun listOut(c: UInt, values: String): Sequence<String> = sequence {
    if(c == 0u) { return@sequence }
    if(c == 1u) { yieldAll(values.asSequence().map{ "$it" }) }
    else {
        values.forEach { ch ->
            yieldAll(listOut(c - 1u, values).map { "$ch$it" })
        }
    }
}

fun String.allUnique(): Boolean {
    this.forEachIndexed { index, ch ->
        this.forEachIndexed { index_, ch_ ->
            if(index_ == index)
                return@forEachIndexed
            if(ch == ch_)
                return false
        }
    }
    return true
}

//; set(v) { this.w = v.x; this.x = v.y; this.x = v.z; this.w = v.w }
fun permutations(vector: String, forms: UInt, format: String, altFormat: String?) = listOut(forms, vector).forEach {
    val build = StringBuilder()
    val unique = it.allUnique()
    val map = (0 until vector.length).map { vector[it] to (altFormat?.get(it) ?: vector[it]) }.toMap()

    fun StringBuilder.appendNam(string: String) = apply { string.forEach { append(map[it]) } }

    build.append("va").append(if(unique) 'r' else 'l').append(" Vector").append(vector.length).append(format).append('.')
        .appendNam(it).append(" get() = Vector").append(it.length).append(format).append('(')
    var notFirst = false
    it.forEach {
        if(notFirst)
            build.append(", ")
        build.append("this.").append(it)
        notFirst = true
    }
    build.append(")")
    if(!unique) { println(build.toString()); return@forEach }
    build.append("; set(v) { ")
    notFirst = false
    it.forEachIndexed { index, ch ->
        if(notFirst)
            build.append("; ")
        build.append("this.").append(ch).append(" = v.").append(vector[index])
        notFirst = true
    }
    build.append(" }")
    println(build.toString())
}

val vec = "xyzw"
val vecC = "rgba"

permutations(vec, 2u, "f", vecC)
println()
permutations(vec, 3u, "f", vecC)
println()
permutations(vec, 4u, "f", vecC)
```

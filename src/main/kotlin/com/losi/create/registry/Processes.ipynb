```kotlin
import com.losi.create.registry.RegisterOrder
import java.util.UUID

object Acts{
    val id1 = UUID.randomUUID()
    val id2 = UUID.randomUUID()
    val id3 = UUID.randomUUID()
    val id4 = UUID.randomUUID()
    val id5 = UUID.randomUUID()
    val id6 = UUID.randomUUID()
    val id7 = UUID.randomUUID()
    val id8 = UUID.randomUUID()

    fun act1() { println("Action A - 1") }
    fun act2() { println("Action B - 2") }
    fun act3() { println("Action C - 3") }
    fun act4() { println("Action D - 4") }
    fun act5() { println("Action E - 5") }
    fun act6() { println("Action F - 6") }
    fun act7() { println("Action G - 7") }
    fun act8() { println("Action H - 8") }
}



RegisterOrder.registerProcess(Acts.id1, Acts::act1) { it.doLast() }
RegisterOrder.registerProcess(Acts.id2, Acts::act2)
RegisterOrder.registerProcess(Acts.id3, Acts::act3)
RegisterOrder.registerProcess(Acts.id4, Acts::act4) { it.justAfter(Acts.id2) }
RegisterOrder.registerProcess(Acts.id5, Acts::act5) { it.justAfter(Acts.id7); it.dependsOn(Acts.id8) }
RegisterOrder.registerProcess(Acts.id6, Acts::act6)
RegisterOrder.registerProcess(Acts.id7, Acts::act7)
RegisterOrder.registerProcess(Acts.id8, Acts::act8)

RegisterOrder.precesses().forEach { it.first.run() }
```
